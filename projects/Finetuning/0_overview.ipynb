# Fine-tuning a Robot Policy

In this course you will take a real-robot vision-language-action model, MolmoAct2, and adapt it to a new skill with a short LoRA fine-tune, then drive the fine-tuned policy live in the LIBERO simulator. Everything runs in the browser on a single AMD Strix Halo machine, so nothing you train or run leaves for a cloud endpoint or a multi-node cluster.

## Overview

1. [Fine-tune MolmoAct2 on LIBERO](finetune_molmoact2_libero.ipynb) - load the DROID base model, sanity-check it open-loop on the real data it was trained on, run a LoRA fine-tune onto the LIBERO simulator, and evaluate the fine-tuned policy closed-loop.
2. [Interactive simulation](interactive_sim_molmoact2_libero.ipynb) - command the fine-tuned policy in natural language and watch it act in the LIBERO simulator, rendered inline in the notebook.
3. [FastWAM world model](inference_fastwam_libero.ipynb) - drive LIBERO with a world-action model and watch it *imagine* the future video, not just the next actions.

## Outcomes

After you've completed this set of materials you will understand how LoRA adapts a large vision-language-action model cheaply enough to fine-tune on a single iGPU, the real-to-sim transfer story that makes the base a good starting point, and how to evaluate and drive a fine-tuned robot policy in a physics simulator. You will have run the whole loop - load, sanity check, fine-tune, evaluate - on AMD ROCm hardware, and you will know which knobs to turn to go from a workshop smoke-test to real training.
